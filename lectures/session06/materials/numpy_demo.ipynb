# Numpy recap: Quick demo

```python
import numpy as np
```

Create a numpy array:

```python
twodim = np.random.random((30, 5))
onedim = np.arange(1,20,1)
```

```python
twodim
```

Explore attributes:

```python
print(twodim.ndim)
print(twodim.size)
twodim.shape
```

Access elements:

```python
onedim[2]
```

```python
twodim[(1,1)]
```

Slicing:

```python
onedim[0:10:2]
```

```python
twodim[2:4:,4::]
```

Ufuncs / aggregates

```python
%timeit multiplied = np.multiply(onedim, 2)
```

```python
%timeit multiplied = onedim*2
```

```python
multiplied
```

```python
np.sum(twodim, axis=1)
```

```python
%timeit sum(twodim)
%timeit np.sum(twodim)
```
