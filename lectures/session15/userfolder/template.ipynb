```python
df = pd.read_csv("patients_results_messy.csv")
df
```

```python
# Clean up column names using string methods
df.columns = df.columns.str.lower().str.replace(r"[^a-z0-9_]+ ", "_", regex=True).str.strip(" ")
df
```

```python
# Split blood presure values into two columns, using the string method split():

df[["systolic", "diastolic"]] = df["blood_pressure"].str.split("/", expand=True)

```

```python
# Converse glucose values

df["glucose"] = pd.to_numeric(df["glucose"], errors="coerce")
df
```

```python
# Standardizing gender 
df["gender"] = df["gender"].str.upper().map(("MALE" : "M", "FEMALE" : "F", "F" : "F", "M" : "M"))
df
```

```python
# Parse date to standard format
# standardize 
df['visit_date_clean'] = pd.to_datetime(df)
```

```python
# Fill or flag missing data (no deletions)
df["flag4inspection"] = df["glucose"].isna()
df
```

```python
df["isagemissing"] = df.age.isna() #isna() makes a boolean series, True if missing, False if not
df
```
