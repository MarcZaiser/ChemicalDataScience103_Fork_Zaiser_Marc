# Python Recap: Binaries and float precision

Computer use binaries, hence, for floats we face certain precision problems: 

```python
0.2+0.3 == 0.5
```

Write a function comparing the sum of two floats `a` with the mathematical result `b` using the absolute difference `abs()` and a tolerance of `1e-9`.

```python
def is_close_enough(a,b,tolerance):
    return abs(a-b) < tolerance
```

```python
is_close_enough(0.1+0.2, 0.3, 1e-09)
```

The same functionality is already implemented as `isclose()` in the `math` package:

```python
import math
math.isclose() #function already pre-existing, returns boolean if two numbers are close to each other within a tolerance
```

# Python Recap: Classes

Classes serve as blueprints for creating objects, bundling together their data and functions.

```python
class Atom:
    def __init__(self, species: str): #links the class Atom to everything that is defined here, linking our object to the class
        self.species = species # define the attributes, here attribute "species"

    def __str__(self): # informal string representation for print() and str()
        return f"Object is defined as {self.species}"
    
    def mass(self) -> float: 
        # average atomic weights from Wikipedia
        if self.species == "H":
            return 1.008
        elif self.species == "C":
            return 12.011
        elif self.species == "O":
            return 15.999
        else:
            raise ValueError(f"Unknown species: {self.species}") # important: error handling! Here: Handle exceptions

# Class basically a blueprint for creating an object; expecially used in DSA104
```

Creating an instance of the Atom class specifying their attributes:

```python
hydrogen = Atom("H")
```

Explore class methods and attributes

```python
print(hydrogen) #printing the object hydrogen gives the upper defined definition of hydrogen
```

```python
hydrogen.mass() #return the defined mass of this object with "object.method()"
```

```python
hydrogen.species #return defined species for here hydrogen
```
