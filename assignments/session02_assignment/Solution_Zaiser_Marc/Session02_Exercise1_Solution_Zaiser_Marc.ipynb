## Exercise 1: Molecular Formula Parser

**Objective:** Write a function `parse_molecular_formula(formula)` that counts atoms in molecular formulas. Start with the simple cases, but design your solution to handle increasingly complex inputs without major rewrites.

**Think computationally:** How can you break this problem into smaller, reusable pieces that won't need to be completely rewritten when the requirements change?


````python
from typing import Dict

def parse_molecular_formula(formula: str) -> Dict[str, int]:
    atom_dict = {}

    i = 0                                                                   # Index that shows/counts position of a character in the string with first element being at index 0
    while i < len(formula):
        # 1) Read the element symbol
        if formula[i].isupper():                                            # If the character at index i is an uppercase letter, it indicates the start of an element symbol
            element = formula[i]                                            # "element" is a new variable that collects the name of the chemical element, that we are "building". Put character at index i into the variable "element"
            i += 1                                                          # Moves one index position forward in the string. Now i=1

            # 2) Check if it is a two-letter element
            if i < len(formula) and formula[i].islower():                   # Now i=1, so we check if the character at index 1 is a lowercase letter. If it is, it indicates that the element symbol has two letters (e.g., "Cl" for Chlorine).
                element += formula[i]                                       # If the character at index 1 is a lowercase letter, we append it to the "element" variable, building the two-letter element symbol. So line says element = element + formula[i]. i itself is not changed here. 
                i += 1                                                      # Moves one index position forward in the string. Now i=2

            # 3) Read the count if it exists
            count_str = ""                                                  # count_str is a new variable to collect the numbers as strings. We collect all numbers as strings first, because we don't know how many digits the number has.
            while i < len(formula) and formula[i].isdigit():                # The while loop continues as long as the character at index i is a digit. Same as with element but with digits. 
                count_str += formula[i]                                     # Append the digit as a string to the count_str 
                i += 1

            count = int(count_str) if count_str else 1                      # Transform string-numbers into integer. If there is no number, we assume the count is 1 (e.g., "H" means "H1").

            # 4) Add to dictionary
            if element in atom_dict:                                        # If the element is already in the dictionary, we add the count to the existing count for that element. This handles cases where the same element appears multiple times in the formula.
                atom_dict[element] += count
            else:                                                           # If element is not in the dictionary, the element is newly added to the dictionary with its count.
                atom_dict[element] = count
        else:
            raise ValueError(f"Invalid character in formula: {formula[i]}") # If the character at index i is not an uppercase letter, it indicates an invalid character in the formula. We raise a ValueError to signal this issue.

    return atom_dict                                                        # Return the dictionary containing the parsed elements and their counts.
````

### Test cases: Set 1 Basic formulae

The test cases will increase in difficulty with each set. Start with the first set, but think about how your approach will scale to the later, more complex cases.

````python
# Simple single-letter elements, single-digit counts
test_cases_level1 = [
    ("H2O", {"H": 2, "O": 1}),
    ("CH4", {"C": 1, "H": 4}),
    ("CO2", {"C": 1, "O": 2}),
    ("NH3", {"N": 1, "H": 3})
]

for formula, expected in test_cases_level1:
    result = parse_molecular_formula(formula)
    print(f"{formula}: {result} {'✓' if result == expected else '✗'}")
````

### Test cases: Set 2 Multi-letter Elements

````python
# Two-letter elements like Ca, Cl, Na
test_cases_level2 = [
    ("NaCl", {"Na": 1, "Cl": 1}),
    ("CaCl2", {"Ca": 1, "Cl": 2}),
    ("MgO", {"Mg": 1, "O": 1}),
    ("AlCl3", {"Al": 1, "Cl": 3}),
    ("H2SO4", {"H": 2, "S": 1, "O": 4})
]

for formula, expected in test_cases_level2:
    result = parse_molecular_formula(formula)
    print(f"{formula}: {result} {'✓' if result == expected else '✗'}")
````

### Test cases: Set 3 Large Numbers

````python
# Multi-digit counts
test_cases_level3 = [
    ("C12H22O11", {"C": 12, "H": 22, "O": 11}),  # Sucrose
    ("Ca10P6O26", {"Ca": 10, "P": 6, "O": 26}),   # Hydroxyapatite
    ("C2H5OH", {"C": 2, "H": 6, "O": 1}),         # Ethanol
    ("Fe2O3", {"Fe": 2, "O": 3})                  # Iron oxide
]

for formula, expected in test_cases_level3:
    result = parse_molecular_formula(formula)
    print(f"{formula}: {result} {'✓' if result == expected else '✗'}")
````
