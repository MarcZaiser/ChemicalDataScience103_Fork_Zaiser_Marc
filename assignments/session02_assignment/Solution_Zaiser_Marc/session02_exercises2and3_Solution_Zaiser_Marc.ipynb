## Exercise 2: Chemical Equation Balancer Checker

**Objective:** Write a program that checks if a chemical equation is balanced.

**Problem:** Create a function called `is_balanced()` that takes a chemical equation string and returns True if balanced, False if not.

**Your Task:**
- Split the equation at the arrow symbol "→" to get reactants and products
- For each side, parse the molecules and their coefficients
- Count total atoms of each element on both sides
- Compare the totals

Example Equation Format: "2H2 + O2 → 2H2O"

**Hints:**

- Use equation.split("→") to separate reactants from products
- Use side.split(" + ") to separate individual molecules
- You'll need to handle coefficients (numbers in front of molecules)
- Reuse your count_atoms() function from Exercise 1!
- For molecules with coefficients like "2H2O", the coefficient applies to all atoms in that molecule

**Steps to Think Through:**

- How do you extract the coefficient from "2H2O"? (It's everything before the first letter)
- How do you multiply the atom counts by the coefficient?
- How do you combine counts from multiple molecules on the same side?


```python
################################# Task 1 #################################
from typing import Dict

def parse_molecular_formula(formula: str) -> Dict[str, int]:
    atom_dict = {}

    i = 0                                                                   # Index that shows/counts position of a character in the string with first element being at index 0
    while i < len(formula):
        # 1) Read the element symbol
        if formula[i].isupper():                                            # If the character at index i is an uppercase letter, it indicates the start of an element symbol
            element = formula[i]                                            # "element" is a new variable that collects the name of the chemical element, that we are "building". Put character at index i into the variable "element"
            i += 1                                                          # Moves one index position forward in the string. Now i=1

            # 2) Check if it is a two-letter element
            if i < len(formula) and formula[i].islower():                   # Now i=1, so we check if the character at index 1 is a lowercase letter. If it is, it indicates that the element symbol has two letters (e.g., "Cl" for Chlorine).
                element += formula[i]                                       # If the character at index 1 is a lowercase letter, we append it to the "element" variable, building the two-letter element symbol. So line says element = element + formula[i]. i itself is not changed here. 
                i += 1                                                      # Moves one index position forward in the string. Now i=2

            # 3) Read the count if it exists
            count_str = ""                                                  # count_str is a new variable to collect the numbers as strings. We collect all numbers as strings first, because we don't know how many digits the number has.
            while i < len(formula) and formula[i].isdigit():                # The while loop continues as long as the character at index i is a digit. Same as with element but with digits. 
                count_str += formula[i]                                     # Append the digit as a string to the count_str 
                i += 1

            count = int(count_str) if count_str else 1                      # Transform string-numbers into integer. If there is no number, we assume the count is 1 (e.g., "H" means "H1").

            # 4) Add to dictionary
            if element in atom_dict:                                        # If the element is already in the dictionary, we add the count to the existing count for that element. This handles cases where the same element appears multiple times in the formula.
                atom_dict[element] += count
            else:                                                           # If element is not in the dictionary, the element is newly added to the dictionary with its count.
                atom_dict[element] = count
        else:
            raise ValueError(f"Invalid character in formula: {formula[i]}") # If the character at index i is not an uppercase letter, it indicates an invalid character in the formula. We raise a ValueError to signal this issue.

    return atom_dict                                                        # Return the dictionary containing the parsed elements and their counts.



################################# Task 2 #################################

def is_balanced(chem_eq: str) -> bool:
    left, right = [part.strip() for part in chem_eq.split("→")]                     # Split the chemical equation into left and right side at arrow, creating a list with two entries being the two sides of the equation. The left side is then assigned to variable "left", the right side to variable "right" and each side is stripped of leading and trailing whitespace
                                                                                

    def count_side(side: str) -> Dict[str, int]:                                    # count_side is a nested function that takes a side of the chemical equation as input and returns a dictionary with the total count of each element on that side 
        total = {}                                                                  # "total" is a dictionary that will store the total count of each element on the given side of the equation like atom_dict before

        molecules = [m.strip() for m in side.split(" + ")]                          # Splits the side of the equation into individual molecules at the " + " sign, creating a list called "molecules" of individual molecule as strings with their coefficients. Each molecule is stripped of leading and trailing whitespace
        for molecule in molecules:                                                  # Iterates over each individual molecule in the list of molecules
            coefficient_str = ""                                                    # coefficient_str is a new variable to collect the numbers as strings (like count_str before)
            j = 0                                                                   # Position index j (like i before)

            while j < len(molecule) and molecule[j].isdigit():                      # While loop that continues as long as j is less than the length of the molecule string and the character at position j is a digit. 
                coefficient_str += molecule[j]                                      # Append the digit as a string to the coefficient_str 
                j += 1

            coefficient = int(coefficient_str) if coefficient_str else 1            # Transform string-numbers into integer. If there is no number, we assume the count is 1 (e.g., "H" means "H1").
            formula_part = molecule[j:]                                             # The rest of the molecule string after the coefficient is extracted as formula_part, which represents the molecular formula without the coefficient

            counts = parse_molecular_formula(formula_part)                          # Uses the previously defined parse_molecular_formula function to get the count of each element in the formula_part without the coefficient, returning a dictionary called "counts"

            for element, amount in counts.items():                                  # count.items() returns the element and its corresponding count (amount) from the counts dictionary with the keys element and the values "amount". 
                total[element] = total.get(element, 0) + amount * coefficient       # .get(element, 0) retrieves the current count of the element in the total dictionary, defaulting to 0 if the element is not yet present. The new count is calculated by adding the product of the amount multiplied by the coefficient to the existing count in total.
                                                                                    # The calculated total count for each element is then stored back in the total dictionary, updating the count for that element.
        return total

    return count_side(left) == count_side(right)
    
```

### Test Cases


```python
# If the outcome is incorrect, the keyword assert will cause an error

print(is_balanced("2H2 + O2 → 2H2O"))
assert is_balanced("2H2 + O2 → 2H2O")

print(is_balanced("H2 + O2 → H2O"))
assert not is_balanced("H2 + O2 → H2O")

print(is_balanced("CH4 + 2O2 → CO2 + 2H2O"))
assert is_balanced("CH4 + 2O2 → CO2 + 2H2O")

print(is_balanced("N2 + H2 → NH3"))
assert not is_balanced("N2 + H2 → NH3")
```

## Exercise 3: Reaction Pathway Analyzer

**Objective:** Analyze multi-step reaction sequences to find synthesis pathways.

**Problem:** Write a program that determines what compounds can be synthesized from starting materials through a series of reactions, and finds a pathway to create a target compound.

**Your Task:**
Create a function called `find_synthesis_path()` that takes:

- A list of starting compounds
- A list of reaction equations
- A target compound to synthesize

The function should return either:

- A list of reaction steps that lead to the target compound, OR
- A message saying the target cannot be synthesized

**Hints:**

- Start by creating a set of "available" compounds (your starting materials)
- For each reaction, check if you have all the reactants needed
- If you can perform a reaction, add the products to your available compounds
- Keep track of which reactions you've used in what order
- Continue until you either find the target or can't make any new compounds

#### Computational Thinking:

**Decomposition:** Break this into smaller problems

- Parse reaction equations
- Track available compounds
- Check if reactions can be performed
- Record the synthesis pathway


**Pattern Recognition:** What patterns do you see?

- Each reaction has the same format: "reactants → products"
- You need to repeatedly check if new reactions become possible


**Algorithm Design:** What's your step-by-step approach?

- Initialize available compounds
- Loop: try each reaction, see if possible, update available compounds
- Stop when target is found or no progress is made


**Abstraction:** What functions might be helpful?

- A function to parse a single reaction equation
- A function to check if a reaction is possible with current compounds
- A function to update available compounds after a reaction



```python
from typing import List
from collections import deque                                                                                   # deque is a double-ended queue that allows for efficient appending and popping from both ends. Each item in the queue stores the current set of available compounds and the path of reactions used so far.

def find_synthesis_path(starting_materials: List[str], reactions: List[str], target: str) -> List[str] | str:   # Find a sequence of reactions that can produce the target compound from the given starting materials.
    
    # 1) Available compounds
    available = set(starting_materials)                                                                         # Create a set of available compounds from the starting materials. Using a set allows for efficient membership testing and avoids duplicates.

    # 2) Queue to test reaction steps one by one
    queue = deque([(set(available), [])])                                                                       # Each item in queue is a tuple containing the current set of available compounds and the path of reactions used so far. The initial state is the starting materials with an empty path.
    visited = {tuple(sorted(available))}                                                                        # visited is a set that keeps track of the states we have already explored to avoid revisiting them.

    # 3) Helper function to split a reaction into reactants and products
    def parse_reaction(reaction: str):
        left, right = [part.strip() for part in reaction.split("→")]                                            # Split the reaction string into left and right parts at the arrow, see Task 2
        
        reactants = [part.strip() for part in left.split("+")]                                                  # Splits the left side of the reaction into individual reactants at the " + " sign, see Task 2
        products = [part.strip() for part in right.split("+")]                                                  # Splits the right side of the reaction into individual products at the " + " sign, see Task 2
    
        reactants = [r.lstrip("0123456789").strip() for r in reactants]                                         # Remove leading coefficients from reactants and products
        products = [p.lstrip("0123456789").strip() for p in products]

        return reactants, products

    # 4) Search until we either find the target or no new compounds can be made
    while queue:
        current_available, path = queue.popleft()
        
        if target in current_available:
            return path

        for reaction in reactions:                                                                              # Iterates over each reaction in the list of reactions to check if it can be applied given the current available compound
            reactants, products = parse_reaction(reaction)

            if all(r in current_available for r in reactants):                                                  # Checks if all reactants for this reaction are available
                new_available = set(current_available)

                for product in products:                                                                        # Adds the products to the available compounds
                    new_available.add(product)
                    
                state = tuple(sorted(new_available))                                                            # Creates a tuple of the new available compounds to represent the state after applying the reaction. 
                if state not in visited:                                                                        # Checks if this new state has already been visited to avoid redundant exploration. If not, it adds the new state to the visited set and appends it to the queue for further exploration.
                    visited.add(state)
                    queue.append((new_available, path + [reaction]))

    # 5) If no path leads to the target
    return f"Target '{target}' cannot be synthesized"
```

### Test Cases

```python
# Basic Multi-Step Synthesis

starting_materials = ["H2", "O2", "N2"]
reactions = [
    "2H2 + O2 → 2H2O",
    "N2 + 3H2 → 2NH3", 
    "NH3 + H2O → NH4OH"
]
target = "NH4OH"

result = find_synthesis_path(starting_materials, reactions, target)
print(result)
# Expected: Should find a path using H2 + N2 → NH3, then NH3 + H2O → NH4OH
```

```python
# Direct Synthesis (One Step)

starting_materials = ["Na", "Cl2"]
reactions = [
    "2Na + Cl2 → 2NaCl",
    "NaCl + H2O → NaOH + HCl"
]
target = "NaCl"

result = find_synthesis_path(starting_materials, reactions, target)
print(result)
# Expected: Should find direct synthesis in one step
```

```python
# Impossible Synthesis

starting_materials = ["H2", "O2"]
reactions = [
    "2H2 + O2 → 2H2O",
    "N2 + 3H2 → 2NH3"  # But we don't have N2!
]
target = "NH3"

result = find_synthesis_path(starting_materials, reactions, target)
print(result)
# Expected: Should return message that NH3 cannot be synthesized
```

```python
# Multiple Pathway Options
starting_materials = ["C", "H2", "O2", "H2O"]
reactions = [
    "C + O2 → CO2",
    "2H2 + O2 → 2H2O",
    "CO2 + H2O → H2CO3",
    "C + 2H2 → CH4",
    "CH4 + 2O2 → CO2 + 2H2O"
]
target = "CO2"

result = find_synthesis_path(starting_materials, reactions, target)
print(result)
# Expected: Should find one of multiple possible paths to CO2
```

```python
# Long Chain Synthesis
starting_materials = ["Fe", "O2", "C"]
reactions = [
    "4Fe + 3O2 → 2Fe2O3",
    "2C + O2 → 2CO",
    "Fe2O3 + 3CO → 2Fe + 3CO2",
    "CO2 + C → 2CO"
]
target = "CO2"

result = find_synthesis_path(starting_materials, reactions, target)
print(result)
# Expected: Should find the multi-step path involving iron oxide formation and reduction
```
